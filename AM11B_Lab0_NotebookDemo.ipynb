## Part 2: What Is a Notebook?

A notebook is a sequence of **cells**. There are two kinds:

- **Markdown cells** (like this one) hold text, explanations, and questions. Double-click one to edit it, then press **Shift + Enter** to show the formatted text again.
- **Code cells** (gray background, with `[ ]` to the left) hold Python code that computes something.

**To run a cell:** click on it, then press **Shift + Enter**. This runs the cell and moves to the next one.

The brackets to the left of a code cell tell you its status:

| You see | It means |
|---|---|
| `[ ]` | Not run yet |
| `[*]` | Running right now. Wait for it to finish. |
| `[3]` | Finished. The number is the order cells were run in. |

**Try it:** click the gray cell below and press **Shift + Enter**.

```python
print("Hello, AM 11B!")
2 + 3
```

You should see the greeting and the answer `5` right below the cell. That's all running a cell is.

**Two rules that save a lot of frustration:**

- **Run cells in order, top to bottom.** Later cells use results from earlier ones. If you see a `NameError`, it almost always means you skipped a cell above. Go back and run it.
- **If things get weird, restart.** Use the menu: **Kernel → Restart Kernel and Run All Cells**. It's the "turn it off and on again" of notebooks, and nothing you do here can break anything permanently.
